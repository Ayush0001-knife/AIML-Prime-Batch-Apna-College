```python
import pandas as pd
```

```python
# Melt ->  To convert data from wide format to long format

# Wide format data
df=pd.DataFrame({
    "country":["USA","USA","India","India"],
    "year":[2020,2021,2020,2021],
    "sales":[100,120,90,110],
    "profit":[20,25,18,22],
})

# Long Format Data
melted_df=df.melt(
    id_vars=["country","year"],             # We want to keep these columns  
    value_vars=["sales","profit"],          # These two columns will be melted into one single column third column
    var_name="metrics",                     # This will be the name of the third column
    value_name="value"                      # This will create the fourth column where the values for the third column will be stored
)

original_df=melted_df.pivot(
    index=["country","year"],               # Here the Original Columns are written
    columns="metrics",                      # Here the name of the column in which melted columns are written
    values="value",                         # Here the values for the columns from the melted column 
)
```
