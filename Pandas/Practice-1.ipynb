```python
import pandas as pd
```

```python
raw_data=pd.read_csv("IRIS.csv")
```

```python
# A=) Display the following information :-> 

# 1-> First 10 rows
raw_data.iloc[0:10]

# 2-> Shape & DataType
raw_data.shape
raw_data.info()

# 3-> Summary statistics (mean,std,min,max)
raw_data.describe()
```

```python
# B=) Select only those rows where: :-> 

# 1-> petal_length>4.5 
raw_data[raw_data["petal_length"]>4.5]

# 2-> species="Iris-virginica" 
raw_data[raw_data["species"]=="Iris-virginica"]
```

```python
# C=) Group by species and compute:
raw_data.groupby("species")

# 1-> Average sepal_length 
raw_data.groupby("species")["sepal_length"].mean()

# 2-> Maximum petal_width
raw_data.groupby("species")["petal_width"].max()

# 3-> Standard deviation of sepal_width
raw_data.groupby("species")["sepal_width"].std()
```

```python
# D=) Create a new column "petal_ratio" = petal_length / petal_width

raw_data["petal_ratio"] = raw_data["petal_length"] / raw_data["petal_width"]

# 1-> For each species find the average of petal ratio
raw_data.groupby("species")["petal_ratio"].mean()
```
