```python
import pandas as pd
```

```python
# Loading Titanic DataSet.

raw_data=pd.read_csv("Titanic-Dataset.csv")
raw_data
```

```python
# A=) Display only columns : Name,Sex,Age,Fare,Survived

raw_data[["Name", "Sex", "Age", "Fare", "Survived"]]
```

```python
# B=) Select passengers who are : female & fare>30

raw_data.query('Sex == "female" & Fare > 30')[["Sex", "Fare"]]
```

```python
# C=) Group by Pclass and compute:
raw_data.groupby("Pclass")

# 1-> Survival rate
raw_data.groupby("Pclass")["Survived"].mean()

# 2-> Average fare
raw_data.groupby("Pclass")["Fare"].mean()

# 3-> Average Age
raw_data.groupby("Pclass")["Age"].mean()
```
