```python
import pandas as pd
```

```python
raw_data=pd.read_csv("raw_data.csv")
raw_data
```

```python
# Transforming Data or Feature Engineering

copy_data=raw_data.copy()
```

```python
# 1->    apply() -> This function is used to add row/column , this function takes another function as the argument and that function whoch is passed as argument can be a logic .

copy_data["tax"]=copy_data["income"].apply(lambda x : "20%" if x>=50000 else "10%")    # logic is passed inside the apply function using lmabda function
copy_data
```

```python
# 2->    map() -> It is used when we have to map existing values with the new values.

# for ex in the data We have to Male->M, Female->F, Unknown-> U

gender_map={"Male":"M","Female":"F","Unknown":"U"}
copy_data["gender"]=copy_data["gender"].map(gender_map)
copy_data
```

```python
# 3 ->      assign() -> To make new changes in the data 

copy_data.assign(new_income=copy_data["income"]*1.1)     # here we have created a new column new_income with exisiting income increased by 10%
```

```python
# 4 ->     replace(old,new) ->   It is used to replace some specific old values with some new values

copy_data["country"]=copy_data["country"].replace("USA","US")
copy_data
```

```python
# 5 ->     rename(columns={old,new}) ->   It is used to rename the rows and columns.

copy_data.columns=["Id","Name","Age","Country","Gender","Income"]  # This is the Raw Way of renaming all the columns

copy_data.rename(columns={"Income":"Salary"})    # This is the way to rename a individual column
copy_data.rename(index={1:"First"})    # This is the way to rename a individual row
```

```python
# 5 ->     sort_values() ->   It is used to sort row/column on the basis of the values in them 

copy_data["Income"].sort_values()  # This will sort the values into ascending order 

# OR    [You can write the above code as like this also]

copy_data.sort_values("Income")  # pandas preserves the original labels

copy_data.sort_values("Income",ascending=False)   # This will sort the values into descending order

sorted_copy_data=copy_data.sort_values(["Income","Age"])   # we can sort the values in two parameters so here sorting will be done on the basis of income but if two incomes are same then sorting will be performed on the basis of Age.
```

```python
# 6 ->     sort_index() ->   It is used to sort index

# But if you will use it on the sorted data it will reverse the effect of sorting and original data wil be recieved.
sorted_copy_data.sort_index()
```

```python
# 7 ->     reset_index() ->   It is used to reset index

# if you use it on the sorted data then that data will remain sorted and the original index will be preserved inside a new column named index but it will create a new label for each row that will start from 0.
sorted_copy_data.reset_index()

sorted_copy_data.reset_index(drop=True)  # Here data will remain sorted and pandas is trying to preserve original index but we don't want original index to be preserved so a new label for each row that will start from 0, without any column having original indices.
```
