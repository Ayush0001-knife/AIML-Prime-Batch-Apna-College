```python
import pandas as pd
```

```python
# merge()   ->   It is used to merge multiple types of dataframes into one dataframe

df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
```

```python
# 1-)  inner join => merge(dataFrame1,dataFrame2,on="") on means parameter which is common between two dataFrames
# The common values of that parameter in both the dataFrames will be there in merged dataFrame

pd.merge(df_customers,df_orders,on="customer_id")
```

```python
# 2-)  left join => We use this when we want full left DataFrame and the common part of the right table in the merged DataFrame.
#  So we use how keyword with the value left at last in merge.

pd.merge(df_customers,df_orders,on="customer_id",how="left")
```

```python
# 3-)  right join => We use this when we want full right DataFrame and the common part of the left table in the merged DataFrame.
#  So we use how keyword with the value right at last in merge.

pd.merge(df_customers,df_orders,on="customer_id",how="right")
```

```python
# 4-)  outer join => We use this when we want data of both the dataFrames in merged DataFrame and the missing values are filled with NaN
#  So we use how keyword with the value outer at last in merge.

pd.merge(df_customers,df_orders,on="customer_id",how="outer")
```
