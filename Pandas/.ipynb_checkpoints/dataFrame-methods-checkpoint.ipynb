```python
import pandas as pd
```

```python
# DataFrame Methods -> These methods help in EDA

data=pd.read_csv("employee-data.csv")
```

```python
# 1-> .head()

data.head()        # It prints data of initial 5 Rows
data.head(2)       # If we want to print only starting 2 rows 
```

```python
# 2-> .tail()

data.tail()        # It prints data of trailing 5 Rows
data.tail(2)       # If we want to print only last 2 rows then 
```

```python
# 3-> .sample() 

data.sample()        # It prints any one random row 
data.sample(4)       # It prints 4 random rows
```

```python
# 3-> .info() 

data.info()            # It gives summary about the Data , how many rows,columns etc etc ....
```

```python
# 4-> .shape

data.shape           # It returns tupple of row and column values
```

```python
# 4-> .describe()

data.describe()       # Gives important statistics of the numerical columns 
```

```python
# 4-> .columns

data.columns       # It wil print the header values of all the columns
```

```python
# 4-> .nynique()

data.nunique()      # It will print the total number of unique values in a column
```
