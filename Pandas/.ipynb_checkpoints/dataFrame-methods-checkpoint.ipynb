```python
import pandas as pd
```

```python
# DataFrame Methods -> These methods help in EDA

data=pd.read_csv("employee-data.csv")
```

```python
# 1-> .head()

data.head()        # It prints data of initial 5 Rows
data.head(2)       # If we want to print only starting 2 rows 
```

```python
# 2-> .tail()

data.tail()        # It prints data of trailing 5 Rows
data.tail(2)       # If we want to print only last 2 rows then 
```

```python
# 3-> .sample() 

data.sample()        # It prints any one random row 
```
