```python
import pandas as pd
```

```python
raw_data=pd.read_csv("raw_data.csv")
raw_data
```

```python
# groupby() -> Here we have to define the parameter on the basis of which we have to group our data.


raw_data.groupby("country")["income"].mean()
raw_data.groupby("country")["income"].max()
raw_data.groupby("country")["income"].sum()
raw_data.groupby("country")["income"].min()
raw_data.groupby("country")["income"].count()


raw_data.groupby("gender")["income"].count()
```
