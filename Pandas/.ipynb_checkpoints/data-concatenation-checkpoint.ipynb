```python
import pandas as pd
```

```python
# Concatenation -> Merging two dataframes either on top of other (row) or side by side (columns) 
```

```python
df1=pd.DataFrame({
    "id":[1,2,3],
    "name":["Adam","Eve","Bob"]
})

df2=pd.DataFrame({
    "id":[4,5,6],
    "name":["Ayush","Luv","Aman"]
})
```

```python
# row-wise Conactenation -> On the Top of the other

pd.concat([df1,df2])                         # Here original labels of each dataframe are preserved
pd.concat([df1,df2],ignore_index=True)       # Here The new serial wise labels will be created 
```

```python
# column-wise Conactenation -> side-by-side of each other

pd.concat([df1,df2],axis=1)
```
