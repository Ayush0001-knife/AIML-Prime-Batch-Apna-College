```python
import pandas as pd
```

```python
# Basic example of Series
# Here Label and Index are both same so index is used
series=pd.Series([1,2,3,4,5])
print(series)
print(type(series))

# Indexing
print(series[0])
```

```python
# Here custom labels are defined 

series=pd.Series([23,24,25,26,27],index=["Ayush","Dishnat","Goldy","Vikas","Sanket"])
print(series)
print(type(series))

# Labelig
print(series["Ayush"])
print(series["Vikas"])
print(series[0])
print(series.index)
```

```python
# Properties of Series 

# 1-> Homogeneous = Data Type will be same no matter what type of data inside series
series1=pd.Series([23,24,25,26,27,"abc"],index=["Ayush","Dishnat","Goldy","Vikas","Sanket",25])
print(series1)
print(type(series1))      

#2 -> Vectorized Operation 
series2=pd.Series([1,2,3,4,5])
series3=pd.Series([6,7,8,9,10])

print(series2 + 2)
print(series3 ** 2)
print(series2+series3)
```
