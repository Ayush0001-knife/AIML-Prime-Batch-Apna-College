```python
import pandas as pd
```

```python
# Basic example of Series
# Here Label and Index are both same so index is used
series=pd.Series([1,2,3,4,5])
print(series)
print(type(series))

# Indexing
print(series[0])
```

```python
# Here custom labels are defined 

series=pd.Series([23,24,25,26,27],index=["Ayush","Dishnat","Goldy","Vikas","Sanket"])
print(series)
print(type(series))

# Labelig
print(series["Ayush"])
print(series["Vikas"])
print(series[0])
print(series.index)
```
