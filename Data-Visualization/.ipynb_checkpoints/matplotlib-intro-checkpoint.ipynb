```python
import matplotlib.pyplot as plt
```

```python
X=[1,2,3,4]
Y=[5,6,7,9]

plt.plot(X,Y)
plt.show()
```

```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M


plt.plot(years,oscar_revenue) 

# Methods to add more info to the graph
plt.title("Oscar movie revenues in each year")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

# label -> it is used To tell which graph line is of what thing

plt.plot(years,oscar_revenue,label="Oscar Movies")
plt.plot(years,non_oscar_revenue,label="Non-Oscar Movies")

plt.title("Oscar Movies Revenue vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")

# legend -> it creates a box in the graph which show different labels inside it 
plt.legend()
plt.legend(loc="center")
plt.legend(loc="lower left")
```

```python
# Format Strings -> It consists of a part of for color,marking and line.

# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

# label -> it is used To tell which graph line is of what thing

# here Below "o--r" -> o represents point , -- represents dottend line , r represents line color to be red 
# here Below "<-y" -> < represents point , - represents normal line , y represents line color to be yellow 

# we can individually change these parameters (marker,line,color)

plt.plot(years,oscar_revenue,"o--r",label="Oscar Movies")
plt.plot(years,non_oscar_revenue,color="pink",label="Non-Oscar Movies")

plt.title("Oscar Movies Revenue vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")

# legend -> it creates a box in the graph which show different labels inside it 
plt.legend()
```

```python
# Other methods and Attributes

plt.plot(years,oscar_revenue,label="Oscar Movies")
plt.plot(years,non_oscar_revenue,label="Non-Oscar Movies")

plt.title("Oscar Movies Revenue vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue in ($M)")

# 1.) style.available  ->  It prints all the available styles in our plot
plt.style.available

# 2.) style.use()  ->  It prints all the available styles in our plot
plt.style.use("dark_background")
plt.style.use("ggplot")
plt.style.use("default")

# 2.) grid()  ->  It is used to display grid in the background of the graph which makes it easier in understanding
plt.grid()
```

```python
# 3.) tight_layout()  ->  When we want to prevent the overlapping between different labels and if we want to give proper sspacing between labels then tight_layout is used.

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]

revenue = [1005, 170, 427, 133, 232]  # in $M

plt.plot(movies, revenue)
plt.title("Movies Revenue")
plt.xlabel("Movies")
plt.ylabel("Revenue ($M)")

plt.tight_layout()

# 4.) savefig("abc.png")  ->  We can save our plot/graph in the png format
plt.savefig("abc.png")
```

```python
# with xkcd() -> To display the graph in the cartoonish way

with plt.xkcd():
     plt.plot(years,oscar_revenue,label="Oscar Movies")
     plt.plot(years,non_oscar_revenue,label="Non-Oscar Movies")

     plt.title("Oscar Movies Revenue vs Non-Oscar Movies Revenue")
     plt.xlabel("Years")
     plt.ylabel("Revenue in ($M)")

     # legend -> it creates a box in the graph which show different labels inside it 
     plt.legend()
```
