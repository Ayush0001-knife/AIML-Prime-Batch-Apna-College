```python
import numpy as np
```

```python
# Integers

arr=np.array([1,2,3,4,5])
print(arr,arr.dtype)
```

```python
# Floating Point Nums

arr1=np.array([1,2,3,4,5.12])
print(arr1,arr1.dtype)

arr2=np.array([1.3,2.4,3.5,4.12,5.12])
print(arr2,arr2.dtype)
```

```python
# String 

arr1=np.array([1,2,3,4,5,"Ayush"])
print(arr1,arr1.dtype)

arr2=np.array(["Hii","I","am","Ayush","Gautam"])
print(arr2,arr2.dtype)
```

```python
# Complex Nums

arr=np.array([2 + 5j])
print(arr,arr.dtype)
```

```python
# Object Data Type

arr=np.array(["Ayush",3,{1,2,5,6}])
print(arr,arr.dtype)
```
