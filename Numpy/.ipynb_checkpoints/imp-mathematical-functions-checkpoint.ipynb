```python
import numpy as np
```

```python
# Creating a sample NumPy array
arr = np.array([1, 2, 3, 4, 5])

print("Array:", arr)
```

```python
# Aggregation Functions

print("Sum:", np.sum(arr))           # Sum of all elements
print("Product:", np.prod(arr))     # Product of all elements
print("Min:", np.min(arr))          # Minimum value
print("Max:", np.max(arr))          # Maximum value
print("Mean:", np.mean(arr))        # Average
print("Standard Deviation:", np.std(arr))  # Spread of data
print("Variance:", np.var(arr))     # Variance
print("Median:", np.median(arr))    # Middle value

print("Index of Min:", np.argmin(arr))  # Index of smallest value
print("Index of Max:", np.argmax(arr))  # Index of largest value
```

```python
# Power Functions

print("Square:", np.square(arr))    # Square of each element
print("Square Root:", np.sqrt(arr)) # Square root

# Power (a^b)
print("Power (arr^2):", np.power(arr, 2))
```

```python
# Rounding Functions

arr2 = np.array([1.2, 2.5, 3.7, 4.1])

print("Original:", arr2)
print("Round:", np.round(arr2))   # Round to nearest integer
print("Floor:", np.floor(arr2))   # Round down
print("Ceil:", np.ceil(arr2))     # Round up
print("Truncate:", np.trunc(arr2)) # Remove decimal part
```

```python
# Log & Exponential Functions

arr3 = np.array([1, 2, 3, 4, 5])

print("Natural Log:", np.log(arr3))     # ln(x)
print("Log base 10:", np.log10(arr3))  # log10(x)
print("Log base 2:", np.log2(arr3))    # log2(x)
print("Exponential:", np.exp(arr3))    # e^x
```

```python
# Extra Functions

arr4 = np.array([1, 2, 2, 3, 4, 4, 5])

print("Unique values:", np.unique(arr4))  # Remove duplicates
print("Sorted array:", np.sort(arr4))    # Sort values
print("Absolute:", np.abs([-1, -2, 3, -4]))  # Absolute values
```
