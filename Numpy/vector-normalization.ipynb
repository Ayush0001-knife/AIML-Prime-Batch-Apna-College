```python
import numpy as np
```

```python
arr=np.array([[1,2],[3,4]])

mean_of_arr=np.mean(arr)        # To find mean of the values of an array
std_of_arr=np.std(arr)          # To find the Standard Deviation

normalized_arr=(arr-mean_of_arr)/std_of_arr          # Formula to Normalize and Array and Mean of this normalized array will be Zero(0) Standard Deviation will be very close to One(1) 
print(normalized_arr)
```
